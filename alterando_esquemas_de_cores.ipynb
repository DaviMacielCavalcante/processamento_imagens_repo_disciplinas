```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
```

```python
arquivo = files.upload()
```

```python
img = cv2.imread('/content/film__3877-jurassic-park--hi_res-458ae800.jpg')
img
```

```python
img_rbg = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
img_hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
img_rbg
```

```python
img_hsv
```

```python
#plt.figure(figsize=(250, 150))
plt.imshow(img_rbg)
plt.title("Imagem em RGB")
plt.axis("off")
plt.show()
```

```python
fig, ax = plt.subplots(1, 3, figsize=(16,4))
ax[0].imshow(img_rbg)
ax[0].set_title("Imagem em RGB")
ax[0].axis("off")
ax[1].imshow(img)
ax[1].set_title("Imagem em BGR")
ax[1].axis("off")
ax[2].imshow(img_hsv)
ax[2].set_title("Imagem em HSV")
ax[2].axis("off")
plt.tight_layout()
plt.show()
```

```python
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
img_gray
```

```python
fig, ax = plt.subplots(1, 4, figsize=(16,4))
ax[0].imshow(img_rbg)
ax[0].set_title("Imagem em RGB")
ax[0].axis("off")
ax[1].imshow(img)
ax[1].set_title("Imagem em BGR")
ax[1].axis("off")
ax[2].imshow(img_hsv)
ax[2].set_title("Imagem em HSV")
ax[2].axis("off")
ax[3].imshow(img_gray, cmap=('gray'))
ax[3].set_title("Imagem em Cinza")
ax[3].axis("off")
plt.tight_layout()
plt.savefig("conversao_de_cores")
plt.show()
```
